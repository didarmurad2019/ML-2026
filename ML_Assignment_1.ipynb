```python
# --- IMPORT LIBRARIES ---
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression

# --- STEP 1: LOAD THE DATASET ---
# Reading the insurance file
df = pd.read_csv('insurance.csv')

# --- STEP 2: DATA CLEANING (Removing Missing Values) ---
# This fixes the "ValueError: Input X contains NaN"
df = df.dropna()

# --- STEP 3: EXPLORATORY DATA ANALYSIS (EDA) ---
print("--- FIRST 5 ROWS OF THE DATASET ---")
print(df.head())

# Visualizing the relationship between Age and Charges
plt.figure(figsize=(10, 6))
sns.regplot(x='age', y='charges', data=df, scatter_kws={'alpha':0.5}, line_kws={'color':'red'})
plt.title('Relationship: Age vs Insurance Charges', fontsize=15)
plt.xlabel('Age of the Person', fontsize=12)
plt.ylabel('Insurance Charges ($)', fontsize=12)
plt.grid(True)
plt.show()

# --- STEP 4: MACHINE LEARNING MODEL (Linear Regression) ---
# Preparing Input (X) and Output (y)
X = df[['age']]
y = df['charges']

# Training the Model
model = LinearRegression()
model.fit(X, y)

# --- STEP 5: RESULTS ---
accuracy = model.score(X, y) * 100
print(f"\nModel Training Successful!")
print(f"Prediction Accuracy (R-squared): {accuracy:.2f}%")
```
